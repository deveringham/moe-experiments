```python
import torch
import time
import h5py
from os import listdir
import lm_eval
from lm_eval.models.huggingface import HFLM
from experiments_pretrained import *
import gc
```

```python
# Set seeds
seed = 42
np.random.seed(seed);
torch.manual_seed(seed);
```

```python
# Experiment parameters
n_samples = 100
max_new_tokens = 100
model_choice = "deepseek"
no_probe = True
```

```python
results = run_experiment_mmlu_eam(model_choice=model_choice,
                                  n_samples=n_samples, start_sample=0,
                                  save_samples=1,
                                  max_new_tokens=max_new_tokens, shuffle_seed=seed,
                                  save_results=False, no_probe=no_probe)
```

```python
def get_avg_inference_time(results):
    time = 0
    n_results = len(results)
    for r in results:
        time += r['metrics']['inference_time']
    return time / n_results
```

```python
avg_time = get_avg_inference_time(results)
print(f"Avg inference time: {avg_time}")
```
